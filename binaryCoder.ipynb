# binary and decimal code interchange module

```python
def convertToBinary(num):
    number = num
    conv = ""
    while number > 0:
        conv+=str(number%2)
        number = number//2
    
    return (conv[::-1])
```

```python
def convertToDecimal(bin):
    sum = 0
    
    l = len(str(bin))
    for i in range(l):
        sum = sum+(bin%10)*2**i
        bin//=10
    return sum
```

```python
x = int(input("1. binary to decimal\n2. decimal to bianry:"))
if x == 1:
    inp = eval(input("enter binary: "))
    for i in inp:
        char = convertToDecimal(i)
        print(chr(char),end="")
elif x==2:
    inp=eval(input("enter in decimal: "))
    for i in inp:
        order = ord(i)
        print(convertToBinary(order),end=",")
```
