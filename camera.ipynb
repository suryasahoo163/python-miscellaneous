```python
import numpy as np
import cv2
```

```python
cap = cv2.VideoCapture(0)

height = int(cap.get(3))
width = int(cap.get(4))
print(height,width)
while True:
   
    if cv2.waitKey(1) == ord('q'):
        cap.release()
        cv2.destroyAllWindows()
        break
    bool,frame = cap.read()
   
    if bool:
        img = cv2.flip(frame,1)
        # img = cv2.line(img,(0,0),(height,width),(255,0,155),2)
        cv2.imshow("my camera",img)
    else:
        print("Camera not found")
```
